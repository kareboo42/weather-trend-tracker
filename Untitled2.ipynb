```python
import requests
from dotenv import load_dotenv
import os

load_dotenv()
API_KEY = os.getenv("OPENWEATHER_API_KEY")

city = "Rancho Cucamonga"
url = "https://api.openweathermap.org/data/2.5/weather"
params = {
    "q": city,
    "appid": API_KEY,
    "units": "imperial"
}

response = requests.get(url, params=params)
print(response.status_code)
print(response.json())
```

```python
data = response.json()

weather_summary = {
    "city": data["name"],
    "temperature": data["main"]["temp"],
    "feels_like": data["main"]["feels_like"],
    "humidity": data["main"]["humidity"],
    "condition": data["weather"][0]["description"],
    "wind_speed": data["wind"]["speed"]
}

print(weather_summary)
```

```python
cities = ["Rancho Cucamonga", "Los Angeles", "San Diego", "San Francisco", "Sacramento"]

weather_data = []

for city in cities:
    params = {
        "q": city,
        "appid": API_KEY,
        "units": "imperial"
    }
    response = requests.get(url, params=params)
    data = response.json()

    weather_summary = {
        "city": data["name"],
        "temperature": data["main"]["temp"],
        "feels_like": data["main"]["feels_like"],
        "humidity": data["main"]["humidity"],
        "condition": data["weather"][0]["description"],
        "wind_speed": data["wind"]["speed"]
    }
    weather_data.append(weather_summary)

print(weather_data)
```

```python
import pandas as pd

df = pd.DataFrame(weather_data)
print(df)

df.to_csv("weather_data.csv", index=False)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(df["city"], df["temperature"])
plt.xlabel("City")
plt.ylabel("Temperature (°F)")
plt.title("Current Temperature by City")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("temperature_chart.png")
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors

# Normalize temperatures to a 0-1 range so we can map them to colors
norm = mcolors.Normalize(vmin=df["temperature"].min(), vmax=df["temperature"].max())
colors = cm.coolwarm(norm(df["temperature"]))

plt.style.use("seaborn-v0_8-whitegrid")
plt.figure(figsize=(8, 5))
bars = plt.bar(df["city"], df["temperature"], color=colors, edgecolor="black", linewidth=0.7)

plt.xlabel("City", fontsize=11)
plt.ylabel("Temperature (°F)", fontsize=11)
plt.title("Current Temperature by City", fontsize=14, fontweight="bold")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig("temperature_chart.png", dpi=150)
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors

norm = mcolors.Normalize(vmin=df["temperature"].min(), vmax=df["temperature"].max())
colors = cm.coolwarm(norm(df["temperature"]))

plt.style.use("seaborn-v0_8-whitegrid")
fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(df["city"], df["temperature"], color=colors, edgecolor="black", linewidth=0.7)

ax.set_xlabel("City", fontsize=11)
ax.set_ylabel("Temperature (°F)", fontsize=11)
ax.set_title("Current Temperature by City", fontsize=14, fontweight="bold")
ax.set_xticklabels(df["city"], rotation=30)

# Add colorbar as legend
sm = cm.ScalarMappable(cmap=cm.coolwarm, norm=norm)
sm.set_array([])
cbar = fig.colorbar(sm, ax=ax)
cbar.set_label("Temperature (°F)", fontsize=10)

plt.tight_layout()
plt.savefig("temperature_chart.png", dpi=150)
plt.show()
```
